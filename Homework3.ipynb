```python
#1-List the order number, order date, customer number and customer name (first and last) of orders
# placed in January 2017 sent to Colorado recipients (OrdState).
txt='''
SELECT OrderTbl.OrdNo, OrderTbl.OrdDate, Customer.CustNo, Customer.CustFirstName, Customer.CustLastName
FROM Customer JOIN OrderTbl ON Customer.CustNo = OrderTbl.CustNo
WHERE OrderTbl.OrdDate >= '2017-01-01' AND  OrderTbl.OrdDate <= '2017-01-31' AND OrderTbl.OrdState = 'CO';


'''
Q1 =  psql.sqldf(txt)
Q1
```

```python
#2-List the order number, order date, and customer number and customer name (first and last) of orders
# placed before January 23, 2017, where customer address (CusState) is in Colorado ('CO')
txt='''
SELECT OrderTbl.OrdNo, OrderTbl.OrdDate, Customer.CustNo, Customer.CustFirstName, Customer.CustLastName
FROM Customer JOIN OrderTbl ON Customer.CustNo = OrderTbl.CustNo
WHERE  OrdDate<'2017-01-23' and CustState='CO'
'''
Q2 =  psql.sqldf(txt)
Q2
```

```python
#3-List the order number, order date, customer number and
#customer name (first and last) of orders placed in January 2017 placed by Colorado customers
#(CustState) but sent to Washington recipients (OrdState).
txt='''
SELECT OrderTbl.OrdNo, OrderTbl.OrdDate, Customer.CustNo, Customer.CustFirstName, Customer.CustLastName
FROM Customer JOIN OrderTbl ON Customer.CustNo = OrderTbl.CustNo
WHERE OrderTbl.OrdDate >= '2017-01-01' AND  OrderTbl.OrdDate <= '2017-01-31' AND Customer.CustState = 'CO' AND OrderTbl.OrdState = 'WA';
'''
Q3 =  psql.sqldf(txt)
Q3
```

```python
#3-List the order number, order date, customer number and
#customer name (first and last) of orders placed in January 2017 placed by Colorado customers
#(CustState) but sent to Washington recipients (OrdState).
txt='''
SELECT OrderTbl.OrdNo, OrderTbl.OrdDate, Customer.CustNo, Customer.CustFirstName, Customer.CustLastName
FROM Customer JOIN OrderTbl ON Customer.CustNo = OrderTbl.CustNo
WHERE OrderTbl.OrdDate >= '2017-01-01' AND  OrderTbl.OrdDate <= '2017-01-31' AND Customer.CustState = 'CO' AND OrderTbl.OrdState = 'WA';
'''
Q3 =  psql.sqldf(txt)
Q3
```

```python
#4-List the customer number, name (first and last), and balance of Washington
#customers who have placed  one or more orders in February 2017.
txt='''
SELECT DISTINCT Customer.CustNo, Customer.CustFirstName, Customer.CustLastName, Customer.CustBal
FROM Customer JOIN OrderTbl ON Customer.CustNo = OrderTbl.CustNo
WHERE OrderTbl.OrdDate >= '2017-02-01' AND  OrderTbl.OrdDate <= '2017-02-28' AND Customer.CustState = 'WA';
'''
Q4 =  psql.sqldf(txt)
Q4
```

```python
#5-List the customer number, name (first and last), and balance of Washington
#customers who have placed  more than 1 order in February 2017.
txt='''
SELECT  Customer.CustNo, Customer.CustFirstName, Customer.CustLastName, Customer.CustBal
FROM Customer JOIN OrderTbl ON Customer.CustNo = OrderTbl.CustNo
WHERE OrderTbl.OrdDate >= '2017-02-01' AND  OrderTbl.OrdDate <= '2017-02-28' AND Customer.CustState = 'WA'
GROUP BY Customer.CustNo, Customer.CustFirstName, Customer.CustLastName, Customer.CustBal
HAVING COUNT(*) > 1;
'''
Q5 =  psql.sqldf(txt)
Q5
```

```python
#6-List product related information if product price is lower than average price
txt='''
SELECT ProdNo, ProdName, ProdMfg, ProdQOH, ProdPrice, ProdNextShipDate
FROM Product
WHERE ProdPrice < (SELECT AVG(ProdPrice) FROM Product);
'''
Q6 =  psql.sqldf(txt)
Q6
```

```python
#7-if product price is lower than average price, find how many times are they ordered?
#List product price as well.
txt='''
SELECT Product.ProdNo, Product.ProdPrice, COUNT(OrdNo) as 'Number of Order'
FROM Product JOIN OrdLine ON Product.ProdNo = OrdLine.ProdNo
WHERE ProdPrice < (SELECT AVG(ProdPrice) FROM Product) GROUP BY Product.ProdNo, Product.ProdPrice;
'''
Q7 =  psql.sqldf(txt)
Q7
```

```python
#8-List the order number and total amount for orders placed each day.

txt='''
SELECT DATE(OrderTbl.OrdDate) AS Date,
       COUNT(DISTINCT OrderTbl.OrdNo)  AS 'Number of Order',
       SUM(Product.ProdPrice * OrdLine.Qty) AS Amount
FROM OrderTbl
JOIN OrdLine
    ON OrderTbl.OrdNo = OrdLine.OrdNo
JOIN Product
    ON OrdLine.ProdNo = Product.ProdNo
GROUP BY DATE(OrderTbl.OrdDate), OrderTbl.OrdNo
ORDER BY DATE(OrderTbl.OrdDate);
'''
Q8 =  psql.sqldf(txt)
Q8
```

```python
#9-What is the total value of product sold by each employee?
#Included internet orders as well
txt='''
SELECT OrderTbl.EmpNo, SUM(Product.ProdPrice * OrdLine.Qty) AS Amount
FROM OrderTbl
JOIN OrdLine ON OrderTbl.OrdNo = OrdLine.OrdNo
JOIN Product ON OrdLine.ProdNo = Product.ProdNo
GROUP BY OrderTbl.EmpNo;

'''
Q9=  psql.sqldf(txt)
Q9
```

```python
#10-List the total commission earned by each employee in each month of 2017.
txt='''
SELECT OrderTbl.EmpNo, strftime('%m', OrdDate) AS Month, SUM(Product.ProdPrice * OrdLine.Qty * Employee.EmpCommRate) AS TotalCommission
FROM OrderTbl
JOIN OrdLine ON OrderTbl.OrdNo = OrdLine.OrdNo
JOIN Product ON OrdLine.ProdNo = Product.ProdNo
JOIN Employee ON Employee.EmpNo = OrderTbl.EmpNo
WHERE OrdDate LIKE '2017%'
GROUP BY OrderTbl.EmpNo, strftime('%m', OrdDate);
'''
Q10 =  psql.sqldf(txt)
Q10
```

```python
#11-List the customer first and last name and order number and employee first name and last name
#who handled that order where order is given in February 2017.
#.If order is internet order, include them as well
#Note:If order is internet order, employee first name and last name will be null
txt='''

SELECT Customer.CustFirstName, Customer.CustLastName, OrderTbl.OrdNo, Employee.EmpFirstName, Employee.EmpLastName
FROM Customer JOIN OrderTbl ON Customer.CustNo = OrderTbl.CustNo
LEFT JOIN Employee ON OrderTbl.EmpNo = Employee.EmpNo
WHERE OrderTbl.OrdDate >= '2017-02-01' AND  OrderTbl.OrdDate <= '2017-02-28'
ORDER BY OrderTbl.OrdNo;
'''
Q11=  psql.sqldf(txt)
Q11
```

```python
#12-List the total commission earned by each employee. Includenemployee first and last name,
#supervisor last name and first name if any.If employee does not have supervisor, supervisor last name and first name will be null.
txt = '''
SELECT Employee.EmpNo,
       Employee.EmpFirstName,
       Employee.EmpLastName,
       SUM(Product.ProdPrice * OrdLine.Qty * Employee.EmpCommRate) AS 'Commission Earned',
       S.EmpFirstName AS 'Supervisor First Name',
       S.EmpLastName AS 'Supervisor Last Name'
FROM Employee
LEFT JOIN Employee S ON Employee.SupEmpNo = S.EmpNo
JOIN OrderTbl ON Employee.EmpNo = OrderTbl.EmpNo
JOIN OrdLine ON OrderTbl.OrdNo = OrdLine.OrdNo
JOIN Product ON OrdLine.ProdNo = Product.ProdNo
GROUP BY Employee.EmpNo;
'''

Q12 = psql.sqldf(txt)
Q12
```

```python
#13- List product information which is not ordered?
#Note: if here is not such product, nothing other than column name will be displayed.
txt = '''
SELECT Product.*
FROM Product
LEFT JOIN OrdLine ON Product.ProdNo = OrdLine.ProdNo
WHERE OrdLine.ProdNo IS NULL;
'''

Q13 = psql.sqldf(txt)
Q13
```

```python
#14-How many times each product are ordered? What is the total value of orders taken by each product?
#Whatis the total quantity ordered for each product?

txt = '''
SELECT Product.ProdNo,
       COUNT(OrdLine.OrdNo) AS 'Times Ordered',
       SUM(Product.ProdPrice * OrdLine.Qty) AS 'Total Value',
       SUM(OrdLine.Qty) AS 'Total Quantity'
FROM Product
JOIN OrdLine ON Product.ProdNo = OrdLine.ProdNo
GROUP BY Product.ProdNo;
'''

Q14 = psql.sqldf(txt)
Q14
```

```python
#15. How many product are there without ProdNextShipDate.
#What is their average price?
txt = '''
SELECT COUNT(*) AS 'Number of Products',
       AVG(ProdPrice) AS 'Average Price'
FROM Product
WHERE ProdNextShipDate IS NULL;
'''

Q15 = psql.sqldf(txt)
Q15
```
